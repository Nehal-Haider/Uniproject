### D9. Loop over a dict

Using `record` from D7, write three loops:

a) print only the **keys** (`for key in record:`)

b) print only the **values** (`.values()`)

c) print both, as `key: value` (`.items()`)

**Expected output for c):**
```
label: R-001
value: 90.0
limit: 100.0
status: OK
```

````python
record = {"label": "R-001", "value": 90.0, "limit": 100.0, "status": "OK"}

# a) Loop over only the keys
print("--- Keys ---")
for key in record:
    print(key)

# b) Loop over only the values
print("\n--- Values ---")
for value in record.values():
    print(value)

# c) Loop over both keys and values
print("\n--- Key: Value Pairs ---")
for key, value in record.items():
    print(f"{key}: {value}")
````

### C1. Load records safely from a file

Write `load_records(path)` with a one-line docstring. It reads `label,value,limit` lines from a file
and returns a **list of dicts**. If the file does not exist, it prints a clear message and returns an
empty list instead of crashing.

Then load `records_sample.txt` and print each record's label and status, as in D16.

**Test it:**

| Call | Expected output |
|---|---|
| `load_records("records_sample.txt")` | the same three lines as D16 |
| `load_records("nope.txt")` | a clear message, then `[]` |

````python
def load_records(path):
    """Read comma-separated records from a file and return a list of dictionaries."""
    records = []
    try:
        with open(path, "r") as file:
            for line in file:
                # Skip empty lines if any exist
                if not line.strip():
                    continue
                label, value, limit = line.strip().split(",")
                records.append({
                    "label": label,
                    "value": float(value),
                    "limit": float(limit)
                })
    except FileNotFoundError:
        print(f"Error: The file '{path}' was not found.")
    return records

# --- Test Case 1: Existing File ---
print("--- Testing with records_sample.txt ---")
loaded_records = load_records("records_sample.txt")

for record in loaded_records:
    if record["value"] > record["limit"]:
        print(f"{record['label']}: OVER LIMIT")
    else:
        print(f"{record['label']}: OK")

# --- Test Case 2: Missing File ---
print("\n--- Testing with nope.txt ---")
missing_records = load_records("nope.txt")
print(missing_records)
````

### C2. Write the results to a file

Using your records from C1, write one line per record to a new file called `results.txt`,
in the form `label,status`. Then read `results.txt` back and print it.

**Expected output:**
```

R-003,OK
```

## Rules

## When it works

2. If you are loading from a file, try it with a file that does not exist. If
   you're on Threshold/Typical, write the crash down — that's next. If you're on

````python

"""
RECORD CHECK

Name : Nehal Haider
Lane : Cyber Security
Date : 14 October 2026
"""

# 1. FUNCTIONS

def check_record(value, limit):
    """Calculate the difference, percentage and status."""
    difference = value - limit
    percent = (value / limit) * 100

    if value > limit:
        status = "OVER LIMIT"
    else:
        status = "WITHIN LIMIT"

    return difference, percent, status


def print_report(label, value, limit, difference, percent, status):
    """Print the report for one record."""
    print("------------------------------")
    print("Source IP:", label)
    print("Failed logins:", value)
    print("Total attempts:", limit)
    print("Difference:", difference)
    print("Percentage:", round(percent, 2), "%")
    print("Status:", status)


# 2. STORE

records = [
    {"label": "10.0.0.5", "value": 12, "limit": 400},
    {"label": "10.0.0.8", "value": 450, "limit": 400},
    {"label": "10.0.0.12", "value": 200, "limit": 250}
]


# 3. PROCESS + OUTPUT

for record in records:
    difference, percent, status = check_record(
        record["value"], record["limit"]
    )

    print_report(
        record["label"],
        record["value"],
        record["limit"],
        difference,
        percent,
        status
    )
````
