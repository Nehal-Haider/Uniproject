### D6. A default value

1. Define `status_of(percent, warning_at=90)` that returns:
   - `"OVER LIMIT"` if `percent` is 100 or more
   - `"WARNING"` if `percent` is `warning_at` or more
   - `"OK"` otherwise
2. Print `status_of(95)`, then `status_of(85)`, then `status_of(85, 80)`.

**Expected output:** `WARNING`, `OK`, `WARNING`

````python
def within_limit(value, limit):
    return value <= limit

# Call the function with arguments in different orders
print(within_limit(87, 100))
print(within_limit(100, 87))

# EXPLANATION: The two results are different because Python matches arguments by position.
# In the first call, 87 goes into 'value' and 100 goes into 'limit' (87 <= 100 is True).
# In the second call, the positions swap: 100 becomes 'value' and 87 becomes 'limit' (100 <= 87 is False).
````

### D7. Keyword arguments

Use `status_of` from D6.

1. Print `status_of(percent=95, warning_at=96)`.
2. Print it again with the two keyword arguments the other way round.
3. Print `status_of(95, warning_at=96)` (one positional, one keyword).

**Expected output:** `OK`, `OK`, `OK`

Now add the line `status_of(percent=95, 96)`, run the cell, read the error, and then delete that line.

````python
# Assuming the status_of function from D6 is defined as follows:
def status_of(percent, warning_at=90):
    if percent >= 100:
        return "OVER LIMIT"
    elif percent >= warning_at:
        return "WARNING"
    else:
        return "OK"

# 1. Keyword arguments in standard order
print(status_of(percent=95, warning_at=96))

# 2. Keyword arguments in reverse order (order doesn't matter when named!)
print(status_of(warning_at=96, percent=95))

# 3. Mixed positional and keyword argument (positional must come first)
print(status_of(95, warning_at=96))
````

### D8. `*args` and `**kwargs`

a) Define `total(*numbers)` that returns the sum of all the numbers passed in. Print `total(1, 2, 3)`,
   `total(10, 20, 30, 40)` and `total()`.

b) Define `configure(**settings)` that loops over `settings.items()` and prints each one as `key = value`.
   Call it with `configure(host="10.0.0.1", port=8080)`.

**Expected output:**
```
6
100
0
host = 10.0.0.1
port = 8080
```

````python
# ==========================================
# a) Using *args for a variable number of positional arguments
# ==========================================
def total(*numbers):
    return sum(numbers)

# Print results with different amounts of numbers passed in
print(total(1, 2, 3))
print(total(10, 20, 30, 40))
print(total())  # Empty tuple sums to 0


# ==========================================
# b) Using **kwargs for a variable number of keyword arguments
# ==========================================
def configure(**settings):
    for key, value in settings.items():
        print(f"{key} = {value}")

# Call the function with specific keyword arguments
configure(host="10.0.0.1", port=8080)
````

````python
# PREDICTION: This will print 42 first (inside the function),
# followed by 21 on the second line (outside the function).

def double(number):
    number = number * 2
    print(number)

number = 21
double(number)

# This prints the global variable 'number'
print(number)

# EXPLANATION: The variable 'number' is still 21 outside the function because parameters
# are local to the function. When we call double(number), Python passes a copy of the
# value (21) into the function's local parameter. Modifying the local 'number' variable
# inside the function does not affect or change the global 'number' variable outside.
````

### C1. Wire two functions together

Using `percent_of` from D3 and `status_of` from D6 **together**, print a status for each of these
three pairs. Do not write the percentage or status logic a second time.

| value | limit | Expected output |
|---|---|---|
| 87 | 100 | `OK` |
| 120 | 100 | `OVER LIMIT` |
| 45 | 100 | `OK` |

````python
# Assuming percent_of from D3 is defined as:
def percent_of(part, total):
    """Return part as a percentage of total."""
    return (part / total) * 100

# Assuming status_of from D6 is defined as:
def status_of(percent, warning_at=90):
    if percent >= 100:
        return "OVER LIMIT"
    elif percent >= warning_at:
        return "WARNING"
    else:
        return "OK"

# The three value/limit pairs to process
pairs = [
    (87, 100),
    (120, 100),
    (45, 100)
]

# Loop through each pair and pass the output of percent_of into status_of
for value, limit in pairs:
    calculated_percentage = percent_of(value, limit)
    status = status_of(calculated_percentage)
    print(status)
````

### C2. How many are over?

Write `count_over(*values, limit=100)` with a one-line docstring. It returns how many of the values
are over the limit.

| Call | Expected output |
|---|---|
| `count_over(87, 120, 45, 101)` | `2` |
| `count_over(87, 120, 45, 101, limit=50)` | `3` |
| `count_over()` | `0` |

````python
def count_over(*values, limit=100):
    """Return the count of values that are strictly greater than the limit."""
    count = 0
    for value in values:
        if value > limit:
            count += 1
    return count

# Test calls matching the expected output table
print(count_over(87, 120, 45, 101))
print(count_over(87, 120, 45, 101, limit=50))
print(count_over())
````
